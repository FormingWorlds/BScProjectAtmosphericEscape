```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
from stellar_cradle import mass_to_luminosity, mass_to_radius, calculate_T_eff, calculate_B_V, mass_to_B_V, solar_mass, solar_radius, solar_luminosity, get_L_x_L_bol, get_L_x, fetch_Jackson_parameters
import matplotlib.ticker as ticker

```

```python
# --- Formatting Constants for High Readability ---
FONT_TITLE = 18
FONT_LABEL = 16
FONT_TICKS = 14
COLOR_LUMINOSITY = '#56B4E9'  # Sky Blue
COLOR_COLOR_IDX  = '#E69F00'  # Orange/Vermilion

# =========================================================================================================================
# Plot 1: Mass vs Luminosity
# =========================================================================================================================
plt.figure(figsize=(9, 6))

masses_plot_full = np.linspace(0.45, 2.1, 100) * solar_mass
log_mass_units = np.log10(masses_plot_full / solar_mass)
log_lum_units  = [np.log10(mass_to_luminosity(m) / solar_luminosity) for m in masses_plot_full]

plt.plot(log_mass_units, log_lum_units, color=COLOR_LUMINOSITY, lw=3)

plt.xlabel(r'$\log_{10}$ Stellar Mass ($M/M_\odot$)', fontsize=FONT_LABEL, labelpad=10)
plt.ylabel(r'$\log_{10}$ Luminosity ($L/L_\odot$)', fontsize=FONT_LABEL, labelpad=10)
plt.title('Mass-Luminosity Relation (Eker et al. 2018)', fontsize=FONT_TITLE, pad=15)

plt.tick_params(axis='both', which='major', labelsize=FONT_TICKS)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# =========================================================================================================================
# Plot 2: Input Mass vs Observed B-V Color Index
# =========================================================================================================================
plt.figure(figsize=(9, 6))

masses_plot_color = np.linspace(0.45, 2.1, 200) * solar_mass
mass_units_color  = masses_plot_color / solar_mass
bv_color_values   = [mass_to_B_V(m) for m in masses_plot_color]

plt.plot(mass_units_color, bv_color_values, color=COLOR_COLOR_IDX, lw=3)

plt.xlabel('Stellar Mass ($M/M_\odot$)', fontsize=FONT_LABEL, labelpad=10)
plt.ylabel('Observed Color Index ($B-V$)', fontsize=FONT_LABEL, labelpad=10)
plt.title('Main Sequence Mass-to-Color Calibration', fontsize=FONT_TITLE, pad=15)

# Invert color axis because smaller magnitudes indicate hotter, bluer stars
plt.gca().invert_yaxis() 

plt.tick_params(axis='both', which='major', labelsize=FONT_TICKS)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
COLOR_RADIUS = '#009E73'  # Bluish Green / Teal

# =========================================================================================================================
# Plot 3: Stellar Mass vs Radius
# =========================================================================================================================
plt.figure(figsize=(9, 6))

# Sweep across your target FGK range (0.45 to 2.1 solar masses)
masses_plot_radius = np.linspace(0.45, 2.1, 200) * solar_mass
mass_units_radius  = masses_plot_radius / solar_mass
radius_values      = [mass_to_radius(m) / solar_radius for m in masses_plot_radius]

plt.plot(mass_units_radius, radius_values, color=COLOR_RADIUS, lw=3)

plt.xlabel('Stellar Mass ($M/M_\odot$)', fontsize=FONT_LABEL, labelpad=10)
plt.ylabel('Stellar Radius ($R/R_\odot$)', fontsize=FONT_LABEL, labelpad=10)
plt.title('Mass-Radius Relation (Giménez & Zamorano 1985)', fontsize=FONT_TITLE, pad=15)

plt.tick_params(axis='both', which='major', labelsize=FONT_TICKS)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Colorblind-friendly high-contrast hex codes (Okabe-Ito palette selection)
COLOR_LUMINOSITY = '#56B4E9'  # Sky Blue (Left Y-axis)
COLOR_RADIUS     = '#009E73'  # Bluish Green / Teal (Right Y-axis)

# =========================================================================================================================
# Plot 4: Dual Y-Axis Plot (Mass vs. Luminosity AND Radius)
# =========================================================================================================================
fig, ax1 = plt.subplots(figsize=(10, 6.5))

# Generate the shared mass array (0.45 to 2.1 solar masses)
masses_plot = np.linspace(0.45, 2.1, 200) * solar_mass
mass_units  = masses_plot / solar_mass

# -------------------------------------------------------------------------------------------------------------------------
# Left Y-Axis: Logarithmic Luminosity
# -------------------------------------------------------------------------------------------------------------------------
log_lum_units = [np.log10(mass_to_luminosity(m) / solar_luminosity) for m in masses_plot]
line1 = ax1.plot(mass_units, log_lum_units, color=COLOR_LUMINOSITY, lw=3, label=r'$\log_{10}$ Luminosity')

ax1.set_xlabel('Stellar Mass ($M/M_\odot$)', fontsize=FONT_LABEL, labelpad=10)
ax1.set_ylabel(r'$\log_{10}$ Luminosity ($L/L_\odot$)', color=COLOR_LUMINOSITY, fontsize=FONT_LABEL, labelpad=10)

# Match the tick mark text color to the curve color for instant visual navigation
ax1.tick_params(axis='y', labelcolor=COLOR_LUMINOSITY, labelsize=FONT_TICKS)
ax1.tick_params(axis='x', labelsize=FONT_TICKS)
ax1.grid(True, linestyle='--', alpha=0.3)

# -------------------------------------------------------------------------------------------------------------------------
# Right Y-Axis: Linear Radius (The 'twinx' magic)
# -------------------------------------------------------------------------------------------------------------------------
ax2 = ax1.twinx()  # Instantiate a second axes that shares the same x-axis

radius_units = [mass_to_radius(m) / solar_radius for m in masses_plot]
line2 = ax2.plot(mass_units, radius_units, color=COLOR_RADIUS, lw=3, linestyle='--', label='Radius')

ax2.set_ylabel('Stellar Radius ($R/R_\odot$)', color=COLOR_RADIUS, fontsize=FONT_LABEL, labelpad=10)
ax2.tick_params(axis='y', labelcolor=COLOR_RADIUS, labelsize=FONT_TICKS)

# -------------------------------------------------------------------------------------------------------------------------
# Title and Unified Legend
# -------------------------------------------------------------------------------------------------------------------------
plt.title('Main Sequence Structural Relations', fontsize=FONT_TITLE, pad=15)

# Combine the legend tracking from both independent axes into one box
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left', fontsize=FONT_LABEL - 2, framealpha=0.9)

plt.tight_layout()
plt.show()
```

```python
COLORS = ['#E69F00', '#56B4E9', '#009E73'] # Vermilion, Sky Blue, Bluish Green

# FIX: Use plt.subplots() so 'ax' is properly defined for your custom tick locators
fig, ax = plt.subplots(figsize=(10, 6.5))

# 1. Create a logarithmic age array spanning from 1.5 Myr to 10 Gyr, same as Jackson et al. (2012)
ages_yr = np.logspace(6.5, 10, 500)

# 2. Select specific B-V values to compare F, G, and K spectral tracks. Took ish middle of the range for each spectral type.
sample_colors = [0.40, 0.67, 1.00]
labels = [f'F-type dwarf ($B-V = 0.40$)', 
          f'G-type dwarf ($B-V = 0.67$)', 
          f'K-type dwarf ($B-V = 1.00$)']

# 3. Compute and plot each color profile
for B_V, label, color in zip(sample_colors, labels, COLORS):
    # Vectorize the check across the numpy age array
    ratio_values = [get_L_x_L_bol(B_V, a) for a in ages_yr]
    ax.plot(ages_yr, ratio_values, color=color, lw=3, label=label)

# 4. Apply high-visibility styling and log scaling
ax.set_xscale('log')
ax.set_yscale('log')

# Define the exact major positions (-2.5, -3.0, -3.5, -4.0, -4.5, -5.0, -5.5, -6.0)
y_tick_powers = np.arange(-6.0, -2.0, 0.5) 
y_tick_values = 10**y_tick_powers

# Define how you want the labels to look textually
y_tick_labels = [f"-{abs(p):.1f}" if p % 1 != 0 else f"-{abs(int(p))}.0" for p in y_tick_powers]

# Apply the custom locators to the Y axis
ax.yaxis.set_major_locator(ticker.FixedLocator(y_tick_values))
ax.yaxis.set_major_formatter(ticker.FixedFormatter(y_tick_labels))

# Set X-axis tick positions to display half numbers in log space (6.5, 7.0, 7.5...)
x_tick_powers = np.arange(6.5, 10.5, 0.5)
ax.xaxis.set_major_locator(ticker.FixedLocator(10**x_tick_powers))
ax.xaxis.set_major_formatter(ticker.FixedFormatter([f"{p:.1f}" for p in x_tick_powers]))

# 5. JOURNAL-STYLE CONFIGURATION (Inward ticks, all-around framing)
ax.tick_params(axis='both', which='major', direction='in', 
               top=True, right=True, length=8, width=1.2, labelsize=FONT_TICKS)

# Set strict plot limits to clean up edge white spaces
ax.set_xlim(10**6.5, 10**10.0)
ax.set_ylim(10**-6.0, 10**-2.5)

# Labels and Spacing
ax.set_xlabel(r'$\log_{10}(\rm age(yr))$', fontsize=FONT_LABEL, labelpad=10)
ax.set_ylabel(r'$\log_{10}(L_{\rm X} / L_{\rm bol})$', fontsize=FONT_LABEL, labelpad=10)
ax.set_title('Stellar X-ray Activity Evolution (Jackson et al. 2012 Framework)', fontsize=FONT_TITLE, pad=15)

# Set gridlines on both log dimensions to make reading data coordinates easy
ax.grid(True, which="major", linestyle='--', alpha=0.3)
ax.legend(loc='lower left', fontsize=FONT_LABEL - 2, framealpha=0.9)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6.5))

# 1. Generate logarithmic age sweep array
ages_yr = np.logspace(6.5, 10, 500)

# 2. Select specific masses (converted internally to kg using your solar_mass variable)
sample_masses_msun = [0.65, 1.00, 1.40]
labels = [r'K-type Dwarf ($0.65\ M_\odot$)',
          r'G-type Dwarf ($1.00\ M_\odot$, Sun-like)',
          r'F-type Dwarf ($1.40\ M_\odot$)']

# 3. Compute and plot the absolute X-ray decay tracking profiles
for mass_msun, label, color in zip(sample_masses_msun, labels, COLORS):
    mass_kg = mass_msun * solar_mass
    
    # Calculate absolute X-ray power in Watts across the timeline
    Lx_watts = [get_L_x(mass_kg, a) for a in ages_yr]
    
    # Convert to log10 space for plotting custom axis half-ticks matching your reference
    log_Lx = np.log10(Lx_watts)
    
    ax.plot(ages_yr, log_Lx, color=color, lw=3, label=label)

# 4. Set scales
ax.set_xscale('log')
# Note: We leave the Y-axis scale as linear under the hood because we already took np.log10(Lx_watts)
# This allows our FixedLocator system to place half-steps (-3.5, -4.0) effortlessly without clipping.

# -------------------------------------------------------------------------------------------------------------------------
# CUSTOM TICK PLACEMENT (Journal Half-Log steps)
# -------------------------------------------------------------------------------------------------------------------------
# X-Axis half-log ticks (6.5, 7.0, 7.5, ... 10.0)
x_tick_powers = np.arange(6.5, 10.5, 0.5)
ax.xaxis.set_major_locator(ticker.FixedLocator(10**x_tick_powers))
ax.xaxis.set_major_formatter(ticker.FixedFormatter([f"{p:.1f}" for p in x_tick_powers]))

# Y-Axis absolute power ticks. Let's automatically calculate appropriate 0.5-step bounds based on your W outputs
# For typical main sequence stars, absolute X-ray power ranges roughly from 10^18 W to 10^24 W.
# Adjust the range values below if your specific mass-to-luminosity scaling shifts the magnitude frame.
y_tick_powers = np.arange(18.0, 25.5, 0.5)
y_tick_labels = [f"{p:.1f}" if p % 1 != 0 else f"{int(p)}.0" for p in y_tick_powers]
ax.yaxis.set_major_locator(ticker.FixedLocator(y_tick_powers))
ax.yaxis.set_major_formatter(ticker.FixedFormatter(y_tick_labels))

# -------------------------------------------------------------------------------------------------------------------------
# INWARD JOURNAL-STYLE CONFIGURATION
# -------------------------------------------------------------------------------------------------------------------------
ax.tick_params(axis='both', which='major', direction='in', 
               top=True, right=True, length=8, width=1.2, labelsize=FONT_TICKS)

# Set strict window constraints to keep data tight within the axes frames
ax.set_xlim(10**6.5, 10**10.0)
ax.set_ylim(18.0, 25.0)  # Caps the plot cleanly. Adjust bounds if your functions output higher/lower ranges.

# Labels and Spacing
ax.set_xlabel(r'$\log_{10}(\rm age(yr))$', fontsize=FONT_LABEL, labelpad=10)
ax.set_ylabel(r'$\log_{10}(L_{\rm X}\ [\rm W])$', fontsize=FONT_LABEL, labelpad=10)
ax.set_title('Absolute Stellar X-ray Luminosity Evolution', fontsize=FONT_TITLE, pad=15)

# Grid and Legend layout
ax.grid(True, which="major", linestyle='--', alpha=0.3)
ax.legend(loc='lower left', fontsize=FONT_LABEL - 2, framealpha=0.9)

plt.tight_layout()
plt.show()
```

```python
ages_yr = np.logspace(6.5, 10, 500)

# FIX: Set the sweep array limits strictly to the valid boundaries found by your diagnostic script
# This guarantees no out-of-bounds ValueError from Jackson et al. (2012)
masses_msun = np.linspace(0.180, 1.772, 500)

# Create a 2D meshgrid where rows/columns map out every combination of mass and age
Age_Grid, Mass_Grid = np.meshgrid(ages_yr, masses_msun)

# =========================================================================================================================
# 2. Compute the 2D Absolute X-ray Luminosity Matrix
# =========================================================================================================================
Log_Lx_Matrix = np.zeros(Age_Grid.shape)

for i in range(len(masses_msun)):
    m_kg = masses_msun[i] * solar_mass
    for j in range(len(ages_yr)):
        a_yr = ages_yr[j]
        
        # Calculate Lx and take log10 to map out order-of-magnitude changes smoothly
        Lx_watts = get_L_x(m_kg, a_yr)
        Log_Lx_Matrix[i, j] = np.log10(Lx_watts)

# =========================================================================================================================
# 3. Visualization Pipeline (2D Continuous Landscape)
# =========================================================================================================================
fig, ax = plt.subplots(figsize=(11, 7))

# Create the filled contour map.
contour = ax.pcolormesh(Age_Grid, Mass_Grid, Log_Lx_Matrix, cmap='viridis', shading='auto')

# Set X-axis to logarithmic for age evolution
ax.set_xscale('log')

# -------------------------------------------------------------------------------------------------------------------------
# CUSTOM TICK PLACEMENT (Matching your half-number journal style)
# -------------------------------------------------------------------------------------------------------------------------
# X-Axis half-log ticks (6.5, 7.0, 7.5, ... 10.0)
x_tick_powers = np.arange(6.5, 10.5, 0.5)
ax.xaxis.set_major_locator(ticker.FixedLocator(10**x_tick_powers))
ax.xaxis.set_major_formatter(ticker.FixedFormatter([f"{p:.1f}" for p in x_tick_powers]))

# Y-Axis linear mass ticks every 0.2 Solar Masses
ax.yaxis.set_major_locator(ticker.MultipleLocator(0.2))

# -------------------------------------------------------------------------------------------------------------------------
# INWARD JOURNAL-STYLE CONFIGURATION
# -------------------------------------------------------------------------------------------------------------------------
ax.tick_params(axis='both', which='major', direction='in', 
               top=True, right=True, length=8, width=1.2, labelsize=FONT_TICKS)

# FIX: Set the frame boundaries to match the exact boundaries of your generated data array
# This completely eliminates both out-of-bounds crashes and unpainted white areas!
ax.set_xlim(10**6.5, 10**10.0)
ax.set_ylim(0.180, 1.772)

# Labels and Titles with clean LaTeX formatting
ax.set_xlabel(r'$\log_{10}(\mathrm{age}\ [\mathrm{yr}])$', fontsize=FONT_LABEL, labelpad=10)
ax.set_ylabel(r'Stellar Mass ($M / M_{\odot}$)', fontsize=FONT_LABEL, labelpad=10)
ax.set_title(r'Absolute X-ray Luminosity Sweep: $L_{\mathrm{X}}(M, t)$', fontsize=FONT_TITLE, pad=15)

# -------------------------------------------------------------------------------------------------------------------------
# Adding the Colorbar
# -------------------------------------------------------------------------------------------------------------------------
cbar = fig.colorbar(contour, ax=ax, pad=0.02)
cbar.set_label(r'$\log_{10}(L_{\mathrm{X}}\ [\mathrm{W}])$', fontsize=FONT_LABEL, labelpad=12)
cbar.ax.tick_params(labelsize=FONT_TICKS)

plt.tight_layout()
plt.show()
```
