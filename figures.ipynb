```python
import numpy as np
import matplotlib.pyplot as plt
```

#### Figure 14 recreation

```python
from radius_cutoffs import *

from figure14 import r, MT_Mp, r_min, r_cap, r_gi

plt.figure(figsize=(10,8))
plt.plot(r, MT_Mp, color='navy')
plt.xlabel('r [m]')
plt.axvline(r_min, label=r'$r_{min}$', color='tomato', linestyle='--', alpha=0.7)
plt.axvline(r_cap, label=r'$r_{cap}$', color='orange', linestyle='-.', alpha=0.7)
plt.axvline(r_gi,label=r'$r_{gi}$' , color='turquoise', linestyle=':', alpha=0.7)
plt.ylabel(r'$M_T/M_{earth}$')
plt.title(r'Total impactor mass $M_T$ needed to eject atmosphere as function of impactor radius r (fig 14)')
plt.xscale('log')
plt.yscale('log')
plt.legend()
# plt.savefig('Mtotal_vs_impactor_rad')
plt.show()
```

#### Ejected per impactor mass shown for radius based on fig 14

```python
from ejected_mass import M_ejected

M_ej = M_ejected(r_min, r)

plt.figure(figsize=(10,8))
plt.plot(r, M_ej, color='navy')
plt.xlabel('r [m]')
plt.axvline(r_min, label=r'$r_{min}$', color='tomato', linestyle='--', alpha=0.7)
plt.axvline(r_cap, label=r'$r_{cap}$', color='orange', linestyle='-.', alpha=0.7)
plt.axvline(r_gi,label=r'$r_{gi}$' , color='turquoise', linestyle=':', alpha=0.7)
plt.ylabel(r'$M_{eject}/m_{imp}$')
plt.title(r'Ejected mass as function of impactor radius r (fig 14)')
plt.xscale('log')
plt.yscale('log')
plt.legend()
# plt.savefig('Ejected_mass_vs_rad')
plt.show()

```

#### Global mass loss: adiabatic vs. isothermal

```python
from giant_impacts_fractional_loss import x, X_loss

gamma_adi = 4/3
gamma_iso = 5/3

X_loss_adia = X_loss(x, gamma_adi)
X_loss_iso = X_loss(x, gamma_iso)

plt.figure(figsize=(10,8))
plt.plot(x, X_loss_iso, color='navy', label='Isothermal global mass loss')
plt.plot(x, X_loss_adia, color='orange', label='Adiabatic global mass loss')
plt.xlabel(r'$v_{imp}m/v_{esc}M$')
plt.ylabel(r'$X_{loss}$')
plt.title(r'Fractional global mass loss $X_{loss}$')
plt.legend()
# plt.savefig('Mtotal_vs_impactor_rad')
plt.show()
```
