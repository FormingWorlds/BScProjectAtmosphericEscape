```python
import numpy as np

import matplotlib.pyplot as plt
```

```python
# x = (v_imp*m) / (v_esc*M)

x = np.linspace(0.0,1.0,100)

X_loss_iso = 0.4*x + 1.4*(x**2) - 0.8*(x**3)

X_loss_adia = 0.4*x + 1.8*(x**2) - 1.2*(x**3)
```

```python
plt.figure(figsize=(10,8))
plt.plot(x, X_loss_iso, color='navy', label='Isothermal global mass loss')
plt.plot(x, X_loss_adia, color='orange', label='Adiabatic global mass loss')
plt.xlabel(r'$v_{imp}m/v_{esc}M$')
plt.ylabel(r'$X_{loss}$')
plt.title(r'Global mass loss $X_{loss}$')
plt.legend()
# plt.savefig('Mtotal_vs_impactor_rad')
plt.show()
```
